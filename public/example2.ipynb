```python
import numpy as np
import matplotlib.pyplot as plt
```

### f(x)=3x^3 + 2x^2 + 1

```python
# Create x values
x = np.linspace(-10, 10, 400)

# Define the function
y = 3*x**3 + 2*x**2 + 1

# Create the plot
plt.plot(x, y)
plt.title("Plot of f(x) = 3x^4 + 2x^2 + 1")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.ylim(-10,10)
plt.grid(True)

# Show the plot
plt.show()
```
